## Dataset

The dataset contains employee information such as:
- Employee Name
- Salary
- Department
- Experience
- City

```python
import pandas as pd

data = {
    "Employee_Name": [" Rohit Sharma", "virat kohli ", "RAHUL", " Rohit Sharma", "shubman gill"],
    "Salary": [25000, 45000, None, 25000, 60000],
    "Department": ["IT", "it", "HR ", "IT", None],
    "Experience": [2, 5, 3, 2, None],
    "City": [" Jaipur ", "Delhi", " jaipur", "Mumbai", "DELHI"]
}

df = pd.DataFrame(data)

df
```
